```python
import cv2
import numpy as np

# Load the painting and the mask
painting = cv2.imread('/Users/althafali/Downloads/masks/sa/p.jpg')
mask = cv2.imread('/Users/althafali/Downloads/masks/sa/m.png', cv2.IMREAD_GRAYSCALE)

# --- FIX: Resize mask to match the painting ---
# painting.shape returns (height, width, channels)
height, width = painting.shape[:2]
mask = cv2.resize(mask, (width, height), interpolation=cv2.INTER_NEAREST)
# ----------------------------------------------

# 1. Ensure mask is binary (0 or 255)
_, binary_mask = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)

# 2. Invert the mask
mask_inv = cv2.bitwise_not(binary_mask)

# 3. Convert mask to 3 channels to match the painting (BGR)
mask_inv_3ch = cv2.cvtColor(mask_inv, cv2.COLOR_GRAY2BGR)

# 4. Apply the mask
corrupted_img = cv2.bitwise_and(painting, mask_inv_3ch)

# Save the result
cv2.imwrite('corrupted_painting.png', corrupted_img)
print("Success! Corrupted image created.")
```
